```python
import numpy as np
# Creating arrays
arr = np.array([12,3,6,4,33,64,54,82,84,2,98,87])
ar = np.array([[12,3,6],[4,335,64]])

# eye (identity matrix)
eye_mat = np.eye(3)

# Attributes
print("Size:", ar.size)       # Total elements
print("Dimensions:", ar.ndim) # Number of dimensions
print("Shape:", ar.shape)     # Rows and columns
```

```python
ar = np.array([[12,3,6],[4,335,64]])

# sum of whole array
print("Total sum:", ar.sum())

# add vertically (column-wise)
print("Axis 0 (Vertical):", ar.sum(axis=0))

# add horizontally (row-wise)
print("Axis 1 (Horizontal):", ar.sum(axis=1))
```

```python
# Reshaping (View)
arr = np.array([12, 3, 6, 4, 33, 64, 54, 82, 84, 2, 98, 87])
k = arr.reshape(3, 4)
arr[0] = 100
print("Original array after view change:", arr)
print("Reshaped view k:\n", k)  # Change reflects in original array

# Reshape with -1
ar = np.array([[12, 3, 6], [4, 335, 64]])
print("\nReshape with -1:\n", ar.reshape(-1, 3))

# 3D Reshape (FIXED: 2*1*3 = 6 elements)
print("\n3D Reshape:\n", ar.reshape(2, 1, 3))
```

```python
# Make a copy and then change
arr = np.array([12,3,6,4,33,64,54,82,84,2,98,87])
k = arr.reshape(3,4).copy()
arr[0] = 100 # This won't affect k now
print("Original:", arr)
print("Copy k:", k)

# Ravel
ar = np.array([[12,3,6,3,56,78],[4,335,64,65,78,97]])
demo = ar.ravel()
demo[0] = 100
print("Raveled array:", demo)
```

```python
# Broadcasting
d = np.array([100,23,45,67,89,98,87,65,43,32,56,78])
print("Broadcasting d*3:", d*3) # 3 will be multiplied to each element

# normal list
n = [5,6,7]
print("List n*3:", n*3) # will repeat the given list 3 times

# Masking
mask = (d > 80)
print("Mask:", mask)          # Prints True/False values
print("d[mask]:", d[mask])    # Prints true values
print("d[~mask]:", d[~mask])  # Approximation symbol prints false values

# Compound Masking
a = (d > 70) & (d < 90)
print("Compound mask d[a]:", d[a])
```
